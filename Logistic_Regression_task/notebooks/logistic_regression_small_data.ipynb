## Logistic Regression
Despite its name it is an algorithm designed to train supervised classification models. it require three stages :
 - first, the linear equation calculate the sum of the weighted features plus the bias.
 - Then it tronsforms this value into a probability between 0 and 1 via the sigmoid function.
 - Finally, it uses a threshold (0.5 by default) to make a decision between class 1 and 0 (if p < 0.5 then class 0, if p >= 0.5 then class 1).

During training, the algorithm uses a loss function called Binary Cross entropy (log loss) to optimize the weights and bias using an optimization algorithm such as gradient descent. it also utilizes regularization to discourage large weight and help reduce overfiting.

## Requirements:
Packages: pandas and scikit-learn

Datasets: train_preprocessed.csv and test_preprocessed.csv

```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    confusion_matrix,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    classification_report
)

train = pd.read_csv("train_preprocessed.csv")
test = pd.read_csv("test_preprocessed.csv")

X_train = train.drop("preterm_birth", axis=1)
y_train = train["preterm_birth"]

X_test = test.drop("preterm_birth", axis=1)
y_test = test["preterm_birth"]

# Create model
model = LogisticRegression()

# Train
model.fit(X_train, y_train)

# Predictions
y_pred = model.predict(X_test)

# Evaluation
print("Confusion Matrix:\n[[TN   Fp]\n [FN   TP]]")
print(confusion_matrix(y_test, y_pred))
print(classification_report(y_test, y_pred))
y_proba = model.predict_proba(X_test)[:, 1]
print("ROC-AUC for class 1 (preterm_birth):", roc_auc_score(y_test, y_proba))
```

## Understanding the evaluation metrics:
# confusion matrix:
shows the actual class × the predicted class which gives us four outcomes:
 - True Negative : Model correctly identified a non-preterm birth.
 - False Positive : Model predicted preterm birth when there wasn't one.
 - False Negative : Model missed an actual preterm birth.
 - True Positive : Model correctly detected a preterm birth.

# Accuracy:
Interpretation: Out of all the predictions made by the model, what percentage were correct?
Its limitation: Accuracy can be misleading when the dataset is imbalanced.

# Macro Average:
Macro Average calculates the ordinary arithmetic mean of the metric values for each class.

# Weighted Average:
Weighted Average also calculates the mean of the metrics for all classes, but it takes into account the number of observations in each class.

# Precision:
Interpretation: Among all observations that the model predicted as class 1, what percentage actually belong to class 1?

# Recall: 
Interpretation: Among all observations that actually belong to class 1, what percentage did the model successfully identify?

# F1-score:
Interpretation: The F1-score summarizes how well the model balances precision and recall.

# Support:
is the number of actual observations belonging to each class in the test dataset.

# ROC-AUC:
Interpretation: ROC-AUC measures how well the model distinguishes positive observations from negative observations across different thresholds.
